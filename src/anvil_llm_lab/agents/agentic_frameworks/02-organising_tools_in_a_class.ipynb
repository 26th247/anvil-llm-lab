# Organising Tools in a Class

Instead of manually writing JSON descriptions, we can document our functions with docstrings. Agent frameworks like ToyAIKit (and PydanticAI, LangChain, etc.) can analyze these docstrings and automatically generate the tool schemas.

This isn't unique to ToyAIKit - PydanticAI and other frameworks work the same way. Once you learn this pattern, you can apply it anywhere.

First, let's set up our data and index:



```python
from gitsource import GithubRepositoryDataReader, chunk_documents
from minsearch import AppendableIndex

reader = GithubRepositoryDataReader(
    repo_owner="evidentlyai",
    repo_name="docs",
    allowed_extensions={"md", "mdx"},
)
files = reader.read()

parsed_docs = [doc.parse() for doc in files]
chunked_docs = chunk_documents(parsed_docs, size=3000, step=1500)

index = AppendableIndex(
    text_fields=["title", "description", "content"],
    keyword_fields=["filename"]
)
index.fit(chunked_docs)
```

```python
# Define our tools with proper docstrings and type hints:


def search(query: str) -> list[dict]:
    """
    Search the documentation database for relevant results based on a query string.

    Args:
        query (str): The search query to look up in the index.

    Returns:
        list[dict]: A list of search result objects returned by the index.
    """
    results = index.search(
        query=query,
        num_results=5
    )
    return results

def add_entry(filename, title, description, content):
    """
    Add a new documentation entry to the index.

    Args:
        filename (str): The source filename associated with the entry.
        title (str): The title of the documentation entry.
        description (str): A short description summarizing the entry.
        content (str): The full content of the documentation entry.

    Returns:
        str: A status message indicating the result of the operation ("OK").
    """
    entry = {
        'start': 0,
        'content': content,
        'title': title,
        'description': description,
        'filename': filename,
    }
    index.append(entry)
    return "OK"

    # The docstring describes: - What the function does - What arguments it takes and their types - What it returns
```

```python
instructions = """
You're a documentation assistant. 

Answer the user question using the documentation knowledge base

Make 3 iterations:

1) in the first iteration, perform one search
2) in the second interation, analyze the results from the previous search
   and perform 2 more searches
3) synthesise the results into the output

IMPORTANT: at each step, give an explanation of why you want to perform 
search for this particular search query. It should be 2-3 sentences explaining
the logic of your decision.

Use only facts from the knowledge base when answering.
If you cannot find the answer, inform the user.

Our knowledge base is entirely about Evidently, so you don't need to 
include the word 'evidently' in search results
"""
```

```python
# Now we can add these tools to ToyAIKit without manually defining the JSON schema:

from toyaikit.tools import Tools

agent_tools = Tools()
agent_tools.add_tool(search)
agent_tools.add_tool(add_entry)

agent_tools.get_tools()
```

```python
# Let's also set up our agent:

from openai import OpenAI
from toyaikit.llm import OpenAIClient
from toyaikit.chat.runners import OpenAIResponsesRunner
from toyaikit.chat.interface import IPythonChatInterface
from toyaikit.chat.runners import DisplayingRunnerCallback

llm_client = OpenAIClient(
    model="gpt-4o-mini",
    client=OpenAI()
)

chat_interface = IPythonChatInterface()
runner_callback = DisplayingRunnerCallback(chat_interface=chat_interface)
```

```python
# Create the agent:

agent = OpenAIResponsesRunner(
    tools=agent_tools,
    developer_prompt=instructions,
    chat_interface=chat_interface,
    llm_client=llm_client
)
```

```python
result = agent.loop(
    'how can I create evidently dahsbord?',
    callback=runner_callback
)
```

## Creating a Tools Class

While docstrings help with descriptions, we still have a problem: our functions depend on a global index variable. This is fine for notebooks, but bad practice for production code.

Global variables are problematic because: - They make code harder to test - They create hidden dependencies - They make reuse difficult

The solution is encapsulation - put everything in a class:



```python
class SearchIndexTools:
    def __init__(self, index):
        self.index = index

    def search(self, query: str) -> list[dict]:
        """
        Search the documentation database for relevant results based on a query string.

        Args:
            query (str): The search query to look up in the index.

        Returns:
            list[dict]: A list of search result objects returned by the index.
        """
        results = self.index.search(
            query=query,
            num_results=5
        )
        return results

    def add_entry(self, filename, title, description, content):
        """
        Add a new documentation entry to the index.

        Args:
            filename (str): The source filename associated with the entry.
            title (str): The title of the documentation entry.
            description (str): A short description summarizing the entry.
            content (str): The full content of the documentation entry.

        Returns:
            str: A status message indicating the result of the operation ("OK").
        """
        entry = {
            'start': 0,
            'content': content,
            'title': title,
            'description': description,
            'filename': filename,
        }
        self.index.append(entry)
        return "OK"
```

```python
# Now the index is passed in through the constructor, and the class has no global dependencies. Let's test it:

search_tools_object = SearchIndexTools(index)
results = search_tools_object.search('dashboard')
results
```

## Using the Class with Tools
We can now add methods from this class to our tools.



```python
# First, the manual way:
agent_tools = Tools()
agent_tools.add_tool(search_tools_object.search)
agent_tools.add_tool(search_tools_object.add_entry)

agent_tools.get_tools()
```

```python
# But ToyAIKit has a convenient shortcut for this:

agent_tools = Tools()
agent_tools.add_tools(search_tools_object)

agent_tools.get_tools()

# The add_tools method automatically finds all public methods (those without underscore prefixes) 
# and adds them as tools. I wish more frameworks had this feature.
```

```python
# Use it with our agent:

agent = OpenAIResponsesRunner(
    tools=agent_tools,
    developer_prompt=instructions,
    chat_interface=chat_interface,
    llm_client=llm_client
)

result = agent.loop(
    'how can I create evidently dahsbard?',
    callback=runner_callback
)
```

# Moving to a Separate File

With this approach, we can move our tools class to a separate file. This makes our code: - More organized - Easier to test - Reusable across different projects

You can see the implementation here: search_tools.py

Now we can import and use it:



```python
import search_tools

search_tools_object = search_tools.SearchIndexTools(index)
result = search_tools_object.search('dashboard')
result
```

The file includes: 
- Proper imports (List, Dict, Any from typing) 
- Complete type hints on all methods 
- Detailed docstrings

Type hints matter more for stricter frameworks like PydanticAI. ToyAIKit is flexible and doesn't care much about them, but PydanticAI and similar frameworks rely on type hints when generating function descriptions. They're more strict and want complete type annotations.



## Summary

- Docstrings eliminate the need for manual JSON tool descriptions
- A tools class encapsulates dependencies (no more global variables)
- add_tools automatically adds public methods as tools
- Moving tools to separate files improves organization and testability
